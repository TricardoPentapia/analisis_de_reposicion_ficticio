```python
import pandas as pd
```

```python
ventas = pd.read_csv("ventas.csv", parse_dates=["fecha"])
ventas.head()
```

```python
demanda_prom = (ventas.groupby(["store_id", "producto_id"])["unidades_vendidas"].mean().reset_index(name="promedio de unidades vendidas"))
demanda_prom.head()
```

```python
top_5_prom = (demanda_prom.groupby(["producto_id"])["unidades_vendidas"].min())
```
